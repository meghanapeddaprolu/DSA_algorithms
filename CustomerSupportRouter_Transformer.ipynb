```python
import numpy as np
import math
import spacy
```

```python
query=input("enter your customer support query")
print(query)
```

```python
nlp=spacy.load("en_core_web_sm")
doc=nlp(query)
Vectors=[]
for token in doc:
  Vectors.append(token.vector)
print(Vectors)
```

```python
embedding_matrix=np.array(Vectors)

wq=np.random.rand(96,64)
wk=np.random.rand(96,64)
wv=np.random.rand(96,64)

q=np.dot(embedding_matrix,wq)
k=np.dot(embedding_matrix,wk)
v=np.dot(embedding_matrix,wv)
```

```python
scores=np.dot(q,k.T)
scores=scores/math.sqrt(k.shape[1])
exp_scores = np.exp(scores - np.max(scores, axis=1, keepdims=True))

sum_exp = np.sum(exp_scores, axis=1, keepdims=True)

attention_weights = exp_scores / sum_exp
```

```python
output = np.dot(attention_weights, v)

print(output)
```

```python
print("Embedding Matrix Shape:", embedding_matrix.shape)
print("Query Shape:", q.shape)
print("Key Shape:", k.shape)
print("Value Shape:", v.shape)
print("Scores Shape:", scores.shape)
print("Attention Weights Shape:", attention_weights.shape)
print("Output Shape:", output.shape)
```

```python
tokens = [token.text for token in doc]

prompt = f"""
You are an AI assistant.

Customer Query:
{query}

Tokens:
{tokens}

Attention Weight Matrix:
{attention_weights}

Please answer:

1. What is the customer's main issue?
2. Which words are likely the most important?
3. Which customer support category does this belong to?
4. Explain your reasoning based on the tokens and attention.
"""
```

```python
response = client.models.generate_content(
    model="gemini-flash-latest",
    contents=prompt
)

print(response.text)
```
